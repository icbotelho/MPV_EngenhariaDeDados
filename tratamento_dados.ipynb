Criação da camada silver

```python
%sql

CREATE OR REPLACE TABLE engenharia_de_dados.silver_setor_censitario AS

select A.*, count( distinct b.id_setor_censitario) as quantidade_setores from engenharia_de_dados.bronze_dim_municipio A
left join engenharia_de_dados.bronze_fato_setor B
on A.id_municipio = B.id_municipio
group by all
```

Criação da camada gold   
(agregação de informações)

```python
%sql
SHOW TABLES IN engenharia_de_dados
```

```python
%sql
select
   a.id_municipio,
   a.sigla_uf,
   ROUND(a.populacao / a.area, 2) AS densidade_populacional_total,
   ROUND(a.populacao / a.domicilios, 2) AS moradores_por_domicilio,
   ROUND((a.populacao_indigena / a.populacao) * 100, 2) AS pct_indigena,
   ROUND((a.populacao_quilombola / a.populacao) * 100, 2) AS pct_quilombola,   
   a.quantidade_setores,
   a.razao_sexo,
   ROUND(a.taxa_alfabetizacao,2) ASxa_alfabetizacao,
   a.idade_mediana,
   a.indice_envelhecimento
from
    engenharia_de_dados.silver_setor_censitario as a
group by all
```

```python
%sql

CREATE OR REPLACE TABLE engenharia_de_dados.gold_indicadres_consolidados AS
select
   a.id_municipio,
   a.sigla_uf,
   ROUND(a.populacao / a.area, 2) AS densidade_populacional_total,
   ROUND(a.populacao / a.domicilios, 2) AS moradores_por_domicilio,
   ROUND((a.populacao_indigena / a.populacao) * 100, 2) AS pct_indigena,
   ROUND((a.populacao_quilombola / a.populacao) * 100, 2) AS pct_quilombola,   
   a.quantidade_setores,
   a.razao_sexo,
   ROUND(a.taxa_alfabetizacao,2) ASxa_alfabetizacao,
   a.idade_mediana,
   a.indice_envelhecimento
from
    engenharia_de_dados.silver_setor_censitario as a
group by all
```
